```python
import json
import re
import io

fname="dataset"

def create_dataset(api_key,api_secret,access_token,access_secret):
  with open('%s.csv' % (fname), 'w', encoding="utf-8") as file:
    w = csv.writer(file, delimiter='|')
    w.writerow(['Screen Name', 'Text', 'Replies', 'Retweets','Likes'])

        # For each tweet matching hashtag, write relevant info to the spreadsheet
    for tweet in tweepy.Cursor(api.search_tweets, "Hepatitis", count=200,tweet_mode='extended'lang="id").items(2000):
        w.writerow([tweet.user.screen_name,
                  tweet.full_text.replace('\n',' '),
                  tweet.user.listed_count,
                  tweet.retweet_count,
                  tweet.favorite_count])

if __name__ == '__main__':
  create_dataset(api_key,api_key_secret,access_token,access_token_secret)
```
